# Predict canada's per capita income in year 2020. Use a regression model 

```python
import numpy as np
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('canada_per_capita_income.csv')
```

```python
df.head(10)
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('per capita income (USD')
plt.scatter(df.year, df.income, color='red', marker='*')
```

### separate x and y

```python
x = df.drop('income', axis='columns')
x
```

```python
y = df.income
y
```

## Create linear regression model

```python
model = linear_model.LinearRegression()
model.fit(x, y)
```

## Predict per capita income of year 2020

```python
model.predict([[2020]])
```

## Check if the price is correct by formula

```python
model.coef_
```

```python
model.intercept_
```
